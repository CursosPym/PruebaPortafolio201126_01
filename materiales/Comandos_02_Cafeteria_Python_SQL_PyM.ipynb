````python
# @title Preparación del caso · Celda original 18
def limpiar_ventas_cafe(df_cafe, precios_cafe):
    """Prepara un cierre de ingresos bajo las reglas de la cafetería de esta guía.

    Conserva el original, retira duplicados exactos y reconoce ERROR/UNKNOWN.
    Recupera solo faltantes deducibles; excluye el total irrecuperable.
    Conserva ingresos conocidos aunque falte cantidad, producto o fecha.
    """
    resultado = df_cafe.copy(deep=True)
    resultado.columns = list(map(lambda nombre: nombre.replace(" ", "_"), resultado.columns))
    resultado = resultado.drop_duplicates().copy()
    for columna in resultado.columns:
        resultado[columna] = resultado[columna].map(
            lambda valor: np.nan if valor == "ERROR" or valor == "UNKNOWN" else valor)
    for columna in ["Quantity", "Price_Per_Unit", "Total_Spent"]:
        resultado[columna] = resultado[columna].astype(float)

    resultado["Price_Per_Unit"] = resultado["Price_Per_Unit"].fillna(resultado["Item"].map(precios_cafe))
    resultado["Total_Spent"] = resultado["Total_Spent"].fillna(resultado["Quantity"] * resultado["Price_Per_Unit"])
    resultado["Quantity"] = resultado["Quantity"].fillna(resultado["Total_Spent"] / resultado["Price_Per_Unit"])
    resultado["Price_Per_Unit"] = resultado["Price_Per_Unit"].fillna(resultado["Total_Spent"] / resultado["Quantity"])

    resultado = resultado.dropna(subset=["Total_Spent"]).copy()
    resultado["Quantity"] = resultado["Quantity"].astype("Int64")
    for columna in ["Item", "Payment_Method", "Location"]:
        resultado[columna] = resultado[columna].fillna("Desconocido")
    resultado["Transaction_Date"] = pd.to_datetime(resultado["Transaction_Date"], errors="coerce")
    resultado["Mes"] = resultado["Transaction_Date"].dt.month.astype("Int64")
    return resultado

df_limpio_cafe = limpiar_ventas_cafe(df_crudo_cafe, precios_cafe)
pd.testing.assert_frame_equal(df_limpio_cafe, df_pasos)
pd.testing.assert_frame_equal(df_crudo_cafe, crudo_respaldo)
assert len(df_limpio_cafe) == 13
assert df_limpio_cafe["Transaction_ID"].is_unique
print("La función reproduce los pasos y conserva el original intacto.")
````

# 2. SQL

SQL define la consulta; Python envía su texto con `consultar(...)` y recibe el resultado como DataFrame. Las operaciones de pandas y la conexión ya están catalogadas en la sección anterior.


## Catálogo de comandos SQL del proyecto

| Comando, expresión o concepto | Para qué se usó |
| --- | --- |
| `SELECT / FROM` | Elegir columnas o métricas y declarar la tabla de ventas o del menú. |
| `DISTINCT` | Revisar las modalidades diferentes de consumo. |
| `AS / alias de tablas y columnas` | Nombrar indicadores y distinguir tablas o subconsultas con v, m y r. |
| `tabla.columna` | Expresar explícitamente el origen de Producto, Ingreso y VentaID en las uniones. |
| `ORDER BY / ASC / DESC` | Ordenar muestras, periodos e ingresos, usando un desempate reproducible. |
| `LIMIT 5` | Inspeccionar cinco ventas sin confundir la muestra con el conteo global. |
| `WHERE / AND` | Filtrar ventas por producto o combinar fecha y modalidad. |
| `BETWEEN ... AND ...` | Seleccionar fechas ISO del 1 al 28 de febrero, con ambos límites incluidos. |
| `= / > / <>` | Comparar condiciones y detectar identificadores repetidos o un número de filas distinto de uno. |
| `COUNT(*)` | Contar ventas o filas después de un cruce. |
| `COUNT(Cantidad)` | Contar solo las ventas cuya cantidad está informada. |
| `SUM(Cantidad) / SUM(Ingreso)` | Sumar piezas conocidas e ingreso, que son métricas diferentes. |
| `ROUND(..., 2)` | Redondear ingresos al presentar, manteniendo controles sin redondeos intermedios. |
| `GROUP BY` | Obtener una fila por periodo, producto o identificador para resúmenes y controles. |
| `HAVING COUNT(*) > 1 / HAVING COUNT(*) <> 1` | Conservar grupos que evidencien duplicados o multiplicación de ventas. |
| `Subconsulta (...) en JOIN` | Resumir todas las ventas por producto antes de unir al catálogo del menú. |
| `LEFT JOIN` | Conservar todos los productos del menú o todas las ventas, según qué tabla quede a la izquierda. |
| `INNER JOIN` | Mostrar solo coincidencias y cuantificar ventas e ingresos perdidos por producto desconocido. |
| `ON v.Producto = m.Producto` | Especificar la llave del cruce y hacer visible su origen. |
| `USING (Producto)` | Unir mediante una columna que se llama igual en ambas tablas. |
