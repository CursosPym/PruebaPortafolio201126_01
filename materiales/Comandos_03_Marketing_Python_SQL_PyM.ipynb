# 2. SQL

SQL define la consulta; Python envía su texto con `consultar(...)` y recibe el resultado como DataFrame. Las operaciones de pandas y la conexión ya están catalogadas en la sección anterior.


## Catálogo de comandos SQL del proyecto

| Comando, expresión o concepto | Para qué se usó |
| --- | --- |
| `SELECT` | Definir las columnas o expresiones que devuelve cada consulta. |
| `FROM` | Indicar la tabla o subconsulta que suministra los datos. |
| `AS` | Nombrar columnas calculadas y asignar alias a tablas o subconsultas. |
| `tabla.columna` | Distinguir atributos de hogares, respuestas y catálogo durante los cruces. |
| `DISTINCT` | Obtener valores diferentes y una sola llave por hogar aceptante. |
| `WHERE` | Filtrar hogares, respuestas aceptadas o quejas antes de agrupar. |
| `AND` | Exigir simultáneamente hijos, gasto mínimo y ausencia de queja. |
| `=; >; >=` | Aplicar igualdad y umbrales inclusivos o exclusivos del encargo. |
| `+` | Construir TotalHijos, TotalGasto y TotalComprasCanales dentro de SELECT. |
| `COUNT(*)` | Contar filas del universo indicado: hogares, respuestas, aceptantes o resultados de una unión. |
| `SUM(Aceptada); SUM(TotalGasto)` | Sumar los unos de aceptación o el gasto definido para el nivel correcto. |
| `GROUP BY` | Resumir por ID en controles o por CampanaID en aceptaciones. |
| `HAVING` | Conservar grupos repetidos o campañas que alcanzan al menos cuatro aceptaciones. |
| `ORDER BY` | Ordenar de manera reproducible resultados y desempates. |
| `ASC; DESC` | Indicar orden ascendente por identificador y descendente por gasto o aceptaciones. |
| `LIMIT` | Reducir la muestra exploratoria a cinco filas o la prioridad a dos campañas. |
| `Subconsulta en FROM (...) AS alias` | Calcular variables antes de filtrar y contar o sumar resultados de otra consulta. |
| `Subconsulta dentro de JOIN` | Resumir por campaña o construir una llave única de aceptantes antes de relacionar. |
| `LEFT JOIN` | Conservar las seis campañas, incluida la sexta sin respuestas registradas. |
| `INNER JOIN` | Relacionar hogares con respuestas o con la lista de ID únicos aceptantes. |
| `RIGHT JOIN` | Conservar el catálogo ubicado a la derecha; se ejecuta solo con una versión compatible de SQLite. |
| `ON h.ID = r.ID` | Escribir explícitamente la igualdad de llaves de la unión. |
| `USING (CampanaID)` | Unir mediante la columna de igual nombre en ambos lados. |
