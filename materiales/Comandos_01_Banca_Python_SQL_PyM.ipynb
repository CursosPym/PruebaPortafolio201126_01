En banca preparamos créditos, homologamos empresas y consultamos montos autorizados conocidos. El inventario cubre tanto la transformación con pandas como SQLite, controles, visualización y entrega; SQL reúne las cláusulas y expresiones realmente utilizadas para responder a finanzas y conservar el catálogo completo.

Esta referencia reúne los comandos utilizados en **Caso_de_Uso_01_Banca_Python_SQL_PyM.ipynb**. Incluye el análisis,
las funciones y los operadores, además de las instrucciones de conexión, presentación,
validación y entrega. Cada comando se explica en el contexto de este proyecto.

# 2. SQL

SQL define la consulta; Python envía su texto con `consultar(...)` y recibe el resultado como DataFrame. Las operaciones de pandas y la conexión ya están catalogadas en la sección anterior.


## Catálogo de comandos SQL del proyecto

| Comando, expresión o concepto | Para qué se usó |
| --- | --- |
| `SELECT` | Define las columnas del detalle y los indicadores del reporte. |
| `FROM` | Indica creditos, cat_empresas o una subconsulta como origen. |
| `DISTINCT` | Obtiene las empresas diferentes presentes en los créditos. |
| `AS y alias de tabla/columna` | Nombra métricas como CreditosVigentes y usa c/r para catálogo y resumen. |
| `WHERE` | Filtra créditos antes de agrupar o devolver el detalle. |
| `AND` | Combina empresa, estado y umbral de monto en la selección Walmart. |
| `=, >= y >` | Compara estado/empresa, incluye exactamente el umbral e identifica grupos con duplicados. |
| `ORDER BY` | Ordena resultados y establece desempates reproducibles. |
| `ASC / DESC` | Combina monto descendente con operación o empresa ascendente. |
| `LIMIT` | Limita a cinco registros la primera inspección SQL. |
| `COUNT(*)` | Cuenta filas: créditos o repeticiones por operación. |
| `COUNT(MontoAutorizado)` | Cuenta únicamente los créditos con monto informado. |
| `SUM(MontoAutorizado)` | Calcula el monto autorizado conocido sin inventar los faltantes. |
| `GROUP BY` | Cambia el nivel de la tabla a una fila por operación o empresa. |
| `HAVING` | Filtra grupos por más de una fila o suma conocida de al menos 500,000. |
| `ROUND(expresion, 2)` | Presenta el monto en MDP con dos decimales. |
| `/ 1000000.0` | Convierte MXN a MDP usando un divisor decimal. |
| `Subconsulta en FROM/JOIN: (...) AS r` | Resume primero los créditos vigentes para unir una fila por empresa. |
| `LEFT JOIN` | Conserva las cinco empresas del catálogo, incluso sin coincidencia en vigentes. |
| `INNER JOIN` | Contrasta la cobertura al mantener solo las cuatro empresas con coincidencia. |
| `USING (Empresa)` | Relaciona por la columna homónima de ambas tablas. |
| `c.Empresa / r.MontoConocidoMXN` | Califica la procedencia de una columna mediante el alias de su tabla. |
| `Literales de texto y numéricos` | Representan Walmart, Vigente y los umbrales 250000/500000 dentro de SQL. |
| `NULL: significado en los resultados` | Los datos cargados conservan montos desconocidos y el LEFT JOIN produce vacíos sin coincidencia; no se usa IS NULL en este caso. |
